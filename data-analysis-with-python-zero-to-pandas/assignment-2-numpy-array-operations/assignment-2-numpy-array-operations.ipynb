## Function 1 - np.copy

Copies an array to allows separate handling of each arrays

```python
# Example 1 - working
arr1 = [[1, 2],
        [3, 4.]]

arr2 = [[5, 6, 7],
        [8, 9, 10]]

array1 = np.copy(arr1)

print(*arr1,array1,sep='\n')
```

Illustrating the copy function

```python
# Example 2 - working
arr1 = [[1, 2],
        [3, 4.]]

arr2 = [[5, 6, 7],
        [8, 9, 10]]

array1 = np.copy(arr1)
array2 = arr2

array1[0][1] = 3
array2[0][2] = 8

print(*arr1,array1,arr2,array2,sep='\n')
print(*arr1 == array1, arr2 == array2,sep='\n')
```

Illustrating the difference between equating variables and using the copy function

```python
# Example 3 - breaking (to illustrate when it breaks)
arr1 = [[1, 2],
        [3, 4.]]

arr2 = [[5, 6, 7],
        [8, 9, 10]]

array1, array2 = np.copy(arr1, arr2)

print(array1, array2)
```

The code breaks because we try to pass in two argument in np.copy function and to fix this we need to pass each argument separately.

Use this function when you want to create and edit copies of an array without changing the data in the original array

## Function 3 - np.transpose

This function transposes the position to values in an array

```python
# Example 1 - working
arr1 = [[1, 2],
        [3, 4.]]

array1 = np.transpose(arr1)

print(array1)
array1.shape
```

Simple illustration of how np.transpose works

```python
# Example 2 - working
arr1 = [[1, 2],
        [3, 4.]]

arr2 = [[5, 6, 7],
        [8, 9, 10]]

array1 = np.transpose(arr1, axes=(0, 1))
array2 = np.transpose(arr2, axes=(1, 0))


print(*array1,array2,sep='\n')
print(array1.shape,array2.shape)
```

Transposing values while indicating the specific values for each axis

```python
# Example 3 - breaking (to illustrate when it breaks)
arr3 = [[11, 12, 13],
        [14, 15, 16],
        [17, 18, 19]]

# To fix this, change axes=(0,0) to axes=(1,0) or remove the axes argument entirely.
array3 = np.transpose(arr3, axes=(1, 0))

print(array3)
print(array3.shape)
```
